```python
### lamda function 
lamda function anonymous function for a one time use they take any numbers of arguments but only one expression
```

```python
a=lambda x:x*2
a(2)
```

```python
add=lambda a,b,c,:a+b+c
add(3,3,3)
```

```python
name=lambda a,b:a+' '+b
name("amal","razik")
```

```python
mvalue=lambda x,y : x if x>y else y
mvalue(10,9)
```

```python
mvalue=lambda x,y : max(x,y)
mvalue(10,9)
```
